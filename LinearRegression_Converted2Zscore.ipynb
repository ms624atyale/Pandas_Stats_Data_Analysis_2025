```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Reproducibility
np.random.seed(42)

# Generate synthetic dataset for 10,000 adolescents
n_samples = 10000

# Mean and Covariance for Weight (kg) and Height (cm)
mean = [58, 165]
cov = [[100, 75],   # Var(Weight)=100 (SD=10), Covariance=75
       [75, 64]]     # Var(Height)=64 (SD=8)

weight, height = np.random.multivariate_normal(mean, cov, n_samples).T

# Z-score Normalization
weight_z = (weight - np.mean(weight)) / np.std(weight)
height_z = (height - np.mean(height)) / np.std(height)

# Plotting side-by-side comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 1. Raw Data Plot
axes[0].scatter(weight, height, alpha=0.15, color='tab:blue', s=10)
axes[0].axvline(np.mean(weight), color='red', linestyle='--', linewidth=1.5, label=f'Mean Wt ({np.mean(weight):.1f}kg)')
axes[0].axhline(np.mean(height), color='green', linestyle='--', linewidth=1.5, label=f'Mean Ht ({np.mean(height):.1f}cm)')
axes[0].set_title("1. Raw Data (Original Scale)", fontsize=13, fontweight='bold')
axes[0].set_xlabel("Weight (kg)", fontsize=11)
axes[0].set_ylabel("Height (cm)", fontsize=11)
axes[0].legend(loc='upper left')
axes[0].grid(True, linestyle=':', alpha=0.6)

# 2. Z-score Normalized Plot
axes[1].scatter(weight_z, height_z, alpha=0.15, color='tab:purple', s=10)
axes[1].axvline(0, color='red', linestyle='--', linewidth=1.5, label='Mean Wt ($Z=0$)')
axes[1].axhline(0, color='green', linestyle='--', linewidth=1.5, label='Mean Ht ($Z=0$)')
axes[1].set_title("2. Z-score Normalized Data", fontsize=13, fontweight='bold')
axes[1].set_xlabel("Weight ($Z$-score)", fontsize=11)
axes[1].set_ylabel("Height ($Z$-score)", fontsize=11)
axes[1].set_xlim(-4, 4)
axes[1].set_ylim(-4, 4)
axes[1].legend(loc='upper left')
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
